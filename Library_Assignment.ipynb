## Question 2: Implement Book, BookCopy, Student and Loan

I am using `__init__` to set up the objects. I also use `__str__` so that when I print an object, I get useful information.

```python
class Student:
    def __init__(self, name, reg_no, contact):
        self.name = name
        self.reg_no = reg_no
        self.contact = contact
        self.loans = []

    def __str__(self):
        return f"{self.name} ({self.reg_no})"


class Book:
    catalogue = []

    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.copies = []
        Book.catalogue.append(self)

    def __str__(self):
        return f"{self.title} by {self.author}"

    @classmethod
    def all_titles(cls):
        return [book.title for book in cls.catalogue]

    @staticmethod
    def is_valid_isbn(isbn):
        isbn_number = isbn.replace("-", "")
        return len(isbn_number) in (10, 13)


class BookCopy:
    def __init__(self, copy_id, book):
        self.copy_id = copy_id
        self.book = book
        self.is_available = True
        book.copies.append(self)

    def __str__(self):
        status = "available" if self.is_available else "borrowed"
        return f"{self.copy_id}: {self.book.title} ({status})"


class Loan:
    FINE_PER_DAY = 1000

    def __init__(self, student, copy, borrow_date, due_date):
        self.student = student
        self.copy = copy
        self.borrow_date = borrow_date
        self.due_date = due_date
        self.return_date = None

    def __str__(self):
        return f"{self.student.name} borrowed {self.copy.book.title}"

    def calculate_fine(self, today):
        if today <= self.due_date:
            return 0
        days = (today - self.due_date).days
        return days * self.FINE_PER_DAY
```

## Question 3: Implement `mark_borrowed`, `mark_returned` and a simple borrow flow

For borrowing, I first check whether the copy is available. If it is available, I change its status to borrowed. When the book is returned, I change the status back to available.

```python
def mark_borrowed(self):
    if not self.is_available:
        raise ValueError("This book copy is already borrowed.")
    self.is_available = False


def mark_returned(self):
    self.is_available = True


BookCopy.mark_borrowed = mark_borrowed
BookCopy.mark_returned = mark_returned

if copy1.is_available:
    copy1.mark_borrowed()
    print("Book borrowed successfully.")
else:
    print("Book is not available.")

print(copy1)
```

```python
copy1.mark_returned()
print("Book returned.")
print(copy1)
```

## Question 4: Implement `calculate_fine` using `FINE_PER_DAY = 1000`

The fine is 1,000 UGX for every day the book is late. I use dates so that I can calculate the number of late days.

```python
from datetime import date

borrow_date = date(2026, 9, 1)
due_date = date(2026, 9, 5)
today = date(2026, 9, 8)

loan1 = Loan(student1, copy1, borrow_date, due_date)
fine = loan1.calculate_fine(today)

print("Days late:", (today - due_date).days)
print("Fine: UGX", fine)
```

In this example, the book is 3 days late. Since the fine is 1,000 UGX per day, the total fine is 3,000 UGX.

## Question 6: Challenge

The challenge is to prevent a person from borrowing an unavailable copy and to list all loans for a student.



```python
def borrow_book(student, copy, borrow_date, due_date):
    if not copy.is_available:
        print("Cannot borrow: this copy is already borrowed.")
        return None

    copy.mark_borrowed()
    new_loan = Loan(student, copy, borrow_date, due_date)
    student.loans.append(new_loan)
    print("Borrowing successful.")
    return new_loan


def list_student_loans(student):
    if len(student.loans) == 0:
        print(student.name, "has no loans.")
        return

    print("Loans for", student.name)
    for loan in student.loans:
        status = "returned" if loan.return_date else "not returned"
        print("-", loan.copy.book.title, "-", status)
```
